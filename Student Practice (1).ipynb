```python
import pandas as pd

conn = sqlite3.connect('my_data.db')

cursor = conn.cursor()
```

```python

pd.read_csv('sales.csv').to_sql('sales',conn, if_exists='replace',index=False)
```

# * Task 1: retrieve sale_date, product and price from sales tables *

```python
pd.read_sql_query('''
SELECT sale_date, product, price FROM sales
''', conn)
```

```python
pd.read_csv('hr_employees.csv').to_sql('employees',conn,if_exists = 'replace',index = False)
```

# * Task 2: Find employees in Chicago with more than 10 years of experience using WHERE and AND *

```python
pd.read_sql_query('''
SELECT name, city, years_experience
FROM employees
WHERE years_experience > 10 AND city = 'Chicago' 
''', conn)
```

# * Task 3: Display the three largest properties by square footage using ORDER BY and LIMIT *

```python
pd.read_csv("realestate_listings.csv").to_sql('listings',conn, if_exists = 'replace',index = False)
```

```python
pd.read_sql_query('''
SELECT property_type, sqft, bedrooms
FROM listings 
ORDER BY sqft DESC
LIMIT 3
''',conn)
```

# * Task 4: Calculate total weekly impressions using SUM() and GROUP BY

```python
pd.read_csv("marketing_campaigns.csv").to_sql('campaigns', conn, if_exists = "replace",index = False)
```

```python
pd.read_sql_query('''
SELECT week, 
        SUM(impressions) AS total_weekly_impressions
FROM campaigns
GROUP BY week
ORDER BY week ASC
''',conn)
```

# * Task 5: Find dates when total restaurant sales exceeded $400 *

```python
pd.read_csv("restaurant_orders.csv").to_sql('orders', conn, if_exists = 'replace', index = False) 
```

```python
pd.read_sql_query('''
SELECT order_date, SUM(order_total) AS total_sales
FROM orders 
GROUP BY order_date 
HAVING SUM(order_total) > 400 
ORDER BY order_date
''', conn)
```

# * Task 6: Find the average patient age and longest waiting time for each doctor *

```python
pd.read_csv("clinic_visits.csv").to_sql('visits', conn, if_exists = 'replace', index = False)
```

```python
pd.read_sql_query('''
SELECT doctor,
        AVG(patient_age) AS avg_patient_age,
        MAX(wait_minutes) AS longest_waiting_mintues
FROM visits
GROUP BY doctor
ORDER BY  wait_minutes DESC
''', conn)
```

# * Task 7: Join Customer and Transaction Tables to Calculate Total Transaction Amount by Branch *

```python
pd.read_csv("bank_customers.csv").to_sql("customers", conn, if_exists = 'replace', index = False)
pd.read_csv("bank_transactions.csv").to_sql("transactions", conn, if_exists = 'replace', index = False)
```

```python
pd.read_sql_query('''
SELECT c.branch, 
        SUM(t.amount) AS total_transaction_amount
FROM "customers" c
JOIN "transactions" t
        ON c.customer_id = t.customer_id
GROUP BY c.branch
ORDER BY total_transaction_amount DESC
''', conn)
```
